```python
import numpy as np
import pandas as pd

def add_noise_to_lightcurve(csv_file, output_csv, noise_std_dev):
    """
    Adds Gaussian noise to the flux values in a light curve CSV file.

    Args:
        csv_file (str): Path to the input CSV file.
        output_csv (str): Path to save the output CSV file with added noise.
        noise_std_dev (float): Standard deviation of the Gaussian noise.
    """
    try:
        df = pd.read_csv(csv_file)
    except FileNotFoundError:
        print(f"Error: CSV file '{csv_file}' not found.")
        return
    
    if 'flux' not in df.columns:
        print("Error: 'flux' column not found in the CSV file.")
        return

    noise = np.random.normal(0, noise_std_dev, len(df['flux']))
    df['flux_noisy'] = df['flux'] + noise
    df.to_csv(output_csv, index=False)
    print(f"Light curve with noise saved to '{output_csv}'")

if __name__ == "__main__":
    input_file = 'light_curve.csv'  # Replace with your input CSV file path
    output_file = 'light_curve_noisy.csv' # Replace with your desired output file path
    noise_level = 0.1 # Adjust the noise level as needed

    # Create a sample CSV file for demonstration if it doesn't exist
    try:
        df_test = pd.read_csv(input_file)
    except FileNotFoundError:
        df_test = pd.DataFrame({'time': np.arange(100), 'flux': np.sin(np.arange(100) * 0.1) + 10})
        df_test.to_csv(input_file, index=False)
        print(f"Sample CSV file '{input_file}' created.")

    add_noise_to_lightcurve(input_file, output_file, noise_level)
```
