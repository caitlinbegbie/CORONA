```python
df_n = pd.read_csv('light_curve_42462.csv')

def add_noise_1(folder_path, col1_name, col2_name, residual_x, residual_y):
    """
    Adds two new columns to each CSV file in a folder and saves the modified DataFrames as new CSV files.

    Args:
        folder_path (str): The path to the folder containing the CSV files.
        col1_name (str): The name of the first new column.
        col2_name (str): The name of the second new column.
        residual_x (list): A list of values for the first new column.
        residual_y (list): A list of values for the second new column.
    """
    target_length = len(residual_x)

    if not os.path.exists(folder_path):
        raise FileNotFoundError(f"The folder '{folder_path}' does not exist.")
    
    for filename in os.listdir(folder_path):
        if filename.endswith(".csv"):
            file_path = os.path.join(folder_path, filename)
            try:
                df = pd.read_csv(file_path)
            except pd.errors.EmptyDataError:
                print(f"Warning: '{filename}' is empty and will be skipped.")
                continue

            # Debug: Check the length of the DataFrame and residuals
            print(f"Checking file: {filename}")
            print(f"DataFrame length: {len(df)}, Residual length: {target_length}")

            # If the DataFrame length doesn't match, adjust it
            if len(df) < target_length:
                # Pad the DataFrame with NaN values
                padding_needed = target_length - len(df)
                print(f"Padding {filename} with {padding_needed} NaN rows.")
                nan_series = pd.Series([np.nan] * padding_needed)
                df = pd.concat([df, nan_series], ignore_index=True)
            elif len(df) > target_length:
                # Truncate the DataFrame to match the residual length
                print(f"Truncating {filename} to {target_length} rows.")
                df = df.iloc[:target_length]

            # Now the DataFrame length matches the residuals length
            df[col1_name] = residual_x
            df[col2_name] = residual_y

            # Save the modified DataFrame to a new file
            new_filename = f"noisey1_{filename}"
            new_file_path = os.path.join(folder_path, new_filename)
            df.to_csv(new_file_path, index=False)
            print(f"Noise added to '{new_filename}'")

            try:
                os.remove(file_path)
                print(f"Deleted the original file: {filename}")
            except Exception as e:
                print(f"Failed to delete the original file {filename}: {e}")
```

```python
def scale2(folder_path):
    """
    Scales the mag values
    """

    if not os.path.exists(folder_path):
        raise FileNotFoundError(f"The folder '{folder_path}' does not exist.")
    
    for filename in os.listdir(folder_path):
        if filename.endswith(".csv"):
            file_path = os.path.join(folder_path, filename)
            try:
                df = pd.read_csv(file_path)
            except pd.errors.EmptyDataError:
                print(f"Warning: '{filename}' is empty and will be skipped.")
                continue

            # Debug: shows that we got this far
            print(f"Checking file: {filename}")

            cols_to_scale = [df.columns[1], df.columns[2]]
            for col in cols_to_scale:
                df[col] = pd.to_numeric(df[col], errors='coerce') * 0.95
                
            # Save the modified DataFrame to a new file
            new_filename = f"0.95jds {filename}"
            new_file_path = os.path.join(folder_path, new_filename)
            df.to_csv(new_file_path, index=False)
            print(f" '{new_filename}' scaled by 0.95")

            for col in cols_to_scale:
                df[col] = pd.to_numeric(df[col], errors='coerce') * 1.05

            # Save the modified DataFrame to a new file
            new_filename = f"1.05jds {filename}"
            new_file_path = os.path.join(folder_path, new_filename)
            df.to_csv(new_file_path, index=False)
            print(f" '{new_filename}' scaled by 1.05")

            for col in cols_to_scale:
                df[col] = pd.to_numeric(df[col], errors='coerce') * 0.88

            # Save the modified DataFrame to a new file
            new_filename = f"0.88jds {filename}"
            new_file_path = os.path.join(folder_path, new_filename)
            df.to_csv(new_file_path, index=False)
            print(f" '{new_filename}' scaled by 0.88")

            for col in cols_to_scale:
                df[col] = pd.to_numeric(df[col], errors='coerce') * 1.07

            # Save the modified DataFrame to a new file
            new_filename = f"1.07jds {filename}"
            new_file_path = os.path.join(folder_path, new_filename)
            df.to_csv(new_file_path, index=False)
            print(f" '{new_filename}' scaled by 1.07")

            for col in cols_to_scale:
                df[col] = pd.to_numeric(df[col], errors='coerce') * 1.01

            # Save the modified DataFrame to a new file
            new_filename = f"1.01jds {filename}"
            new_file_path = os.path.join(folder_path, new_filename)
            df.to_csv(new_file_path, index=False)
            print(f" '{new_filename}' scaled by 1.01")

            for col in cols_to_scale:
                df[col] = pd.to_numeric(df[col], errors='coerce') * 0.85

            # Save the modified DataFrame to a new file
            new_filename = f"0.85jds {filename}"
            new_file_path = os.path.join(folder_path, new_filename)
            df.to_csv(new_file_path, index=False)
            print(f" '{new_filename}' scaled by 0.85")
```
