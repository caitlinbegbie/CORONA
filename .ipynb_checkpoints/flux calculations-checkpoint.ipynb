```python
import csv 
import pandas as pd
import numpy as np 
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('help - lc_wise-toi-6 (2).csv')
```

```python
def augment_value(value):
    flux = 10**(-0.4 * (value - 17.8)) 				
    return flux
```

```python
df['flux'] = df['mags'].apply(augment_value)
```
