```python
import csv

x = []
y = []

try:
    with open('help - lc_wise-toi-6 (1).csv', 'r') as csvfile:
        plots = csv.reader(csvfile, delimiter=',')
        
        # Check if the file has rows
        header = next(plots, None)  # Return None if there's no header
        if header is None:
            print("The file is empty or improperly formatted.")
        else:
            for row in plots:
                if row:  # Ensure that the row is not empty
                    x.append(row[0])
                    y.append(row[1])
except FileNotFoundError:
    print("The file was not found.")
except StopIteration:
    print("The file is empty or all rows have been read.")
except Exception as e:
    print(f"An error occurred: {e}")
```

```python
plt.plot(x, y, marker='o') # 'o' specifies circle markers
plt.xlabel('X-axis label')
plt.ylabel('Y-axis label')
plt.title('Plot from CSV')
plt.grid(True)
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the CSV file into a pandas DataFrame
data = pd.read_csv('help - lc_wise-toi-6 (1).csv')

# Ensure time data is in correct format (if needed)
data['time'] = pd.to_datetime(data['jds'])

# Plot the light curve
plt.figure(figsize=(10, 6))  # Adjust figure size as needed
plt.plot(data['time'], data['mags'], marker='o', linestyle='-', markersize=3)
plt.xlabel('Time (years)')
plt.ylabel('Apparent Magnitude')
plt.title('Light Curve for lc_wise-toi-6')
plt.gca().invert_yaxis()  # Invert y-axis if magnitude scale is used
plt.grid(True)

plt.savefig("my_plot.png")
plt.show()
```
