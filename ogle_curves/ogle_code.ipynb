# OGLE Data:
## Convert .tar Files to .csv

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import os
import tarfile
from astropy.table import Table   # format : "ascii"
from pathlib import Path
```

```python

def tar_to_csv(tar_file_path, output_folder = "C:/Users/Home/ulab/CORONA/ogle_curves"):
    open = tar_file_path.extractfile("C:/Users/Home/ulab/CORONA/ogle_curves")  # folder
    csv_filepaths = []
    with tarfile.open(tar_file_path, 'r') as tar:
            tar.extractfile(output_folder)
    for filename in os.listdir(output_folder):
        if filename.endswith(".csv"):
                csv_filepath = os.path.join(output_folder, filename, 'w')
                csv_filepaths.append(csv_filepath)
    tar_file_path.close()
    return csv_filepaths

#use "r" and "w" to index when converting to csv
# with open  . . . as dat_file

tar_to_csv('C:/Users/Home/ulab/Corona/ogle_curves')
```

```python

def tar_csv(input_file_path, output_folder = "C:/Users/Home/ulab/CORONA/ogle_curves"):
    os.makedirs(output_folder, exist_ok=True)

    try:
        with tarfile.open(input_file_path, 'r') as tar:
            found_csv = False
            for member in tar.getmembers():
                if member.isfile() and member.name.endswith('.csv'):
                    found_csv = True
                    with open(output_file_path, 'w') as outfile:
                        outfile.write(tar.extractfile(member).read())
                if not found_csv:
                    print(f"No CSV files found in '{input_file_path}'.")
    except tarfile.ReadError:
        print(f"Error: '{input_file_path}' is not a valid tar file or is corrupted.")
    except FileNotFoundError:
        print(f"Error: The input file '{input_file_path}' was not found.")
    except Exception as e:
        print(f"An unexpected error occurred: {e}")

tar_csv('query_OGLE-LMC-RCB-01.tar')
```
