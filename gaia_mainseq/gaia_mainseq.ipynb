# Convert Main Seq LCs to .csv   from Gaia alerts

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
def add_cols(file_path):
    df = pd.read_csv(file_path, comment='#', names=["Date", "JD (TCB)", "avg Mag"])
    df_print = print(df.head(128))
    return df
```

```python
Gaia18acn = pd.DataFrame(add_cols('Gaia18acn.csv'))
Gaia19ekz = pd.DataFrame(add_cols('Gaia19ekz.csv'))
Gaia19fcp = pd.DataFrame(add_cols('Gaia19fcp.csv'))
Gaia20aoq = pd.DataFrame(add_cols('Gaia20aoq.csv'))
Gaia20bns = pd.DataFrame(add_cols('Gaia20bns.csv'))
Gaia21ack = pd.DataFrame(add_cols('Gaia21ack.csv'))
```

```python
new_Gaia18acn = Gaia18acn.to_csv('new_Gaia18acn.csv', index = False)
new_Gaia19ekz = Gaia19ekz.to_csv('new_Gaia19ekz.csv', index = False)
new_Gaia19fcp = Gaia19fcp.to_csv('new_Gaia19fcp.csv', index = False)
new_Gaia20aoq = Gaia20aoq.to_csv('new_Gaia20aoq.csv', index = False)
new_Gaia20bns = Gaia20bns.to_csv('new_Gaia20bns.csv',index = False)
new_Gaia21ack = Gaia21ack.to_csv('new_Gaia21ack.csv'))
```
